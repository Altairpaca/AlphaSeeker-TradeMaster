```python
for col in df.columns:
    cdf = df[col]
    if cdf.isna().sum() == 0:
        continue
    tnas=cdf.isna().sum()
    head_nas = cdf.first_valid_index()
    tail_nas = len(cdf) - cdf.last_valid_index() - 1
    if tnas == head_nas + tail_nas:
        continue
        pass
    print(f'{col} has {tnas} missing values, \t{head_nas} at head,\t{tnas-head_nas-tail_nas} in the middle.')
```
